```python
from bs4 import BeautifulSoup
import requests
URL  = 'https://timely-sunshine-e821b3.netlify.app/'
Page = requests.get(URL)
Page.status_code
```

```python
htmlCode = Page.text
```

```python
#print(htmlCode)
soup = BeautifulSoup(htmlCode,'html.parser')
print(soup)
```

```python
c = soup.find_all('div',attrs={'class':'main'})
for i in c:
    print("Div: \n",i,end='\n\n\n')
```

```python
content = soup.find('div',attrs={'class':'price'})
t = content.text
print(t)
```

```python
content = soup.find('div',attrs={'class':'price'})
t = content.text.strip()
print(t)
```

```python
content = soup.find('div',attrs={'class':'image'})
t = content.text
print(t)
```

```python
i = content.find('img')
print(i)
```

```python
im = i['src']
print(im)
```

```python
item = soup.find_all('div',class_='a')
names_list = []
prices_list = []
images_list = []

for items in item:
    extracted_name = items.find('div',class_='name').text.strip()
    extracted_price = items.find('div',class_='price').text.strip()
    image_tag = items.find('div',class_='image').find('img')
    extracted_image_src = image_tag['src'] if image_tag else None

    names_list.append(extracted_name)
    prices_list.append(extracted_price)
    images_list.append(extracted_image_src)

print("Names : ", names_list)
print("Prices : ", prices_list)
print("images : ", images_list)
```

```python
df = pd.DataFrame({'Product_Name':names_list,"Product_Prices ":prices_list,"Product_images":images_list})
```
