```python
import matplotlib.pyplot as p
n = int(input("Enter the Hours here : "))
h = [i for i in range(n)]
visitors = [10,20,30,40,10,50,60,600,100]
p.plot(h,visitors,
       color='blue',
       marker='o',
       linestyle = "--")
p.title("Website Visitors per hours")
p.xlabel("Hours per Day")
p.ylabel("Number of Vositors")
p.legend()

p.grid(True,linestyle='-',alpha = 0.3)
p.show()
```

```python
product = ["Laptop","phones","tabs","T.V","Smartwatch"]
sales = [10,30,40,0.5,10]
p.bar(product,sales,color='blue',edgecolor='blue',alpha=0.2,width=1)
p.title("Product Sales")
p.xlabel("Product")
p.ylabel("Sales")
p.show()
```

```python
marks = [23,54,54,34,65,65,34,37,97,78,56,98,0,9]
p.hist(marks,bins=10,histtype='bar',color='purple',edgecolor='blue')
p.title("Distribution of Students")
p.xlabel("marks range")
p.ylabel("Number of Students")
p.show()
```

```python
b = ["apple","Samsung","Lenovo","Lava","Moto"]
P = [30,50,40,1,20]
p.pie(P,labels=b,autopct="%.2f%%",startangle=0
      ,colors=["blue","green","purple","yellow","red"],
      explode = [1,0,0,0,0],
      shadow = True)
p.title("Mobile Market Share 2026")
p.show()
```

```python
import matplotlib.pyplot as p

score = [
    [1, 212, 3, 4, 5, 6, 7],
    [2, 4, 7, 8, 9, 65, 4],
    [2, 8, 5, 42, 654, 8],
    [75, 75, 4, 83, 74, 6, 8]
]

p.boxplot(
    score,
    patch_artist=True,
    boxprops=dict(facecolor='red'),
    medianprops=dict(color='blue', linewidth=2),
    whiskerprops=dict(color='green'),
    capprops=dict(color='black'),
    notch=False,
    vert=True,
    labels=["A", "B", "C", "D"]
)

p.title("Exam Scores Distribution Across Classes")
p.xlabel("Classes")
p.ylabel("Scores")
p.grid(True, linestyle='--', alpha=0.7)

p.show()
```

```python
import matplotlib.pyplot as p
days = ["mon","tue","wed","thur","fry","Sat","Sun"]
visitors = [120,32,54,66,88,11,111]
p.fill_between(days,visitors,
               color = "Skyblue",alpha=0.4,
               label="Website Visitors")
p.plot(days,visitors,color='black')
p.title("Website Visitors Over a Week ")
p.xlabel("Day of Week")
p.ylabel("Visitors")
p.grid(True, linestyle='--', alpha=0.7)

p.show()
```
